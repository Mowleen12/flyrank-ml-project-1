## 2. The question: decision, action, cost of a wrong call

### Search Question
Which content pages should be prioritized for review because they have the highest opportunity for improvement?

### Unit of Analysis
One row represents one content page and its associated historical search and engagement metrics.

### Decision
The SEO or content strategy team must decide which pages should be refreshed before others.

### Output
A ranked **Content Refresh Opportunity Score** with supporting reason codes explaining why each page was prioritized.

### Action
Editors review the highest-ranked pages and decide whether to update titles, improve content quality, refresh outdated information, optimize metadata, or improve internal linking.

### Cost of a Wrong Recommendation

**False Positive:** A page is recommended for review even though it does not require updating, resulting in unnecessary editorial effort.

**False Negative:** A page that genuinely needs improvement is not recommended, causing missed opportunities for increased traffic, engagement, or conversions.

### Why Data and Machine Learning Help
Content performance depends on many signals such as impressions, clicks, click-through rate, rankings, and engagement metrics. A data-driven scoring approach can combine these signals consistently across thousands of pages and help teams prioritize limited resources more effectively than manual review alone.
